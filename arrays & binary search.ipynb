```python
class Solution:
    def sort012(self, arr):
        zeros = 0
        ones = 0
        twos = 0

        for number in arr:
            if number == 0:
                zeros += 1
            elif number == 1:
                ones += 1
            else:
                twos += 1

        arr[:] = [0] * zeros + [1] * ones + [2] * twos
```
